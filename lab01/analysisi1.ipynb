```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np 

result = pd.read_csv("result.csv")
result 
```

```python
result_clean = result.drop(columns=["Арга"])
aos = []
soa = []

for column in result_clean.columns:
    aos.append(result_clean[column].iloc[1].astype("float32"))
    soa.append(result_clean[column].iloc[5].astype("float32"))

x_labels = ["1сая", "10 сая", "50 сая", "100 сая"]
x = np.arange(len(x_labels))
width = 0.35  

plt.figure(figsize=(8, 7))

plt.bar(x - width/2, aos, width, label='AoS') 
plt.bar(x + width/2, soa, width, label='SoA')

plt.title('AoS vs SoA (seq_mass)', fontsize=14, pad=15)

plt.xticks(x, x_labels)

plt.grid(axis='y', color='lightgray', linestyle='-', linewidth=0.8, zorder=0)
plt.gca().set_axisbelow(True)

plt.legend(loc='lower center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=False)


plt.tight_layout()
plt.show()
```
